# Customer Churn Prediction - Exploratory Data Analysis (EDA)

This notebook analyzes customer characteristics, contractual relationships, and billing patterns to uncover drivers of telecom churn.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
fig_dir = Path('../reports/figures')
fig_dir.mkdir(parents=True, exist_ok=True)

df = pd.read_csv('../data/processed/clean.csv')
print('Dataset Shape:', df.shape)
df.head()
```

## 1. Churn Distribution

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=df, x='Churn', hue='Churn', palette=['#3b82f6', '#ef4444'], legend=False)
total = len(df)
for p in ax.patches:
    height = p.get_height()
    pct = height / total * 100
    ax.annotate(f'{int(height)} ({pct:.1f}%)', (p.get_x() + p.get_width() / 2., height / 2),
                ha='center', va='center', color='white', fontweight='bold')
plt.title('Overall Churn Distribution', fontsize=12, fontweight='bold')
plt.xlabel('Churn (0 = Retained, 1 = Churned)')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig(fig_dir / 'churn_distribution.png', dpi=300)
plt.show()
```

## 2. Churn by Contract and Internet Service

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df, x='Contract', hue='Churn', palette=['#3b82f6', '#ef4444'], ax=axes[0])
axes[0].set_title('Churn by Contract Type', fontweight='bold')

sns.countplot(data=df, x='InternetService', hue='Churn', palette=['#3b82f6', '#ef4444'], ax=axes[1])
axes[1].set_title('Churn by Internet Service', fontweight='bold')
plt.tight_layout()
plt.savefig(fig_dir / 'churn_by_contract.png', dpi=300)
plt.show()
```

## 3. Churn by Payment Method

```python
plt.figure(figsize=(9, 4.5))
sns.countplot(data=df, y='PaymentMethod', hue='Churn', palette=['#3b82f6', '#ef4444'])
plt.title('Churn by Payment Method', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig(fig_dir / 'churn_by_payment_method.png', dpi=300)
plt.show()
```

## 4. Tenure and Monthly Charges Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data=df, x='tenure', hue='Churn', kde=True, ax=axes[0], palette=['#3b82f6', '#ef4444'], bins=30)
axes[0].set_title('Tenure Distribution by Churn', fontweight='bold')
sns.histplot(data=df, x='MonthlyCharges', hue='Churn', kde=True, ax=axes[1], palette=['#3b82f6', '#ef4444'], bins=30)
axes[1].set_title('Monthly Charges Distribution by Churn', fontweight='bold')
plt.tight_layout()
plt.savefig(fig_dir / 'tenure_monthlycharges_histograms.png', dpi=300)
plt.show()
```

## 5. Correlation Heatmap

```python
plt.figure(figsize=(6, 5))
num_cols = df.select_dtypes(include=[np.number]).columns
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Heatmap of Numeric Features', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig(fig_dir / 'correlation_heatmap.png', dpi=300)
plt.show()
```
